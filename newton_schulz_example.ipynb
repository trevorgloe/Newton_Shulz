# Newton-Schulz algorithm

The algorithm atempts to find $A^{-1}$ from $A$ and an initial guess, $G_0$, by doing the following iteration
$$
G_k = G_{k-1} + F_{k-1}G_{k-1}
$$
$$
F_k = I - G_k A
$$
So each iteration, we have to compute $2$ matrix-matrix products.

Exact inverse algorithms, (such as $LU$ factorization) have complexity $O(n^3)$ where $n$ is the dimension of the matrix

In our case, we have a sequence of matrices $A_i$ that we want to invert. At each iteration, we have $A_{i-1}^{-1}$ and we want to find $A_i^{-1}$ given that 
$$
A_i = A_{i-1} + P
$$
where $P$ is a small perturbation.

```python
n = 70
A = np.random.randn(n,n)
print(A)
P = 1e-2*np.random.rand(n,n)
A2 = A + P
print(A2)
```

```python
def NS(A, G0, kmax):
    G = np.copy(G0)
    I = np.eye(A.shape[0])
    for k in range(kmax):
        F = I - G @ A
        G = G + F @ G
        print(F)
        
    return G
```

```python
Ai_inv = np.linalg.inv(A)
Ai2_inv = NS(A2, Ai_inv, 7)
```

```python
print(Ai2_inv @ A2)
print(np.linalg.norm(Ai2_inv @ A2 - np.eye(A.shape[0])))
```
