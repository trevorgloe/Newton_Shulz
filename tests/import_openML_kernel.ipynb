# Example of importing openML data for a kernel matrix
Below we use the function in `functions/import_openML_data.py` to create a kernel matrix from the test data

The kernel matrix is defined by 
$$
  K_{i,j} = k(x_i, x_j)
$$
where $x_i$ is the i'th element of the imported dataset.

The function call specifies `m`, which is how many of the data points are sampled. The result is an $m\times m$ matrix.

```python
import sys
from pathlib import Path
import time

import numpy as np

REPO_ROOT = Path.cwd().parent
if str(REPO_ROOT) not in sys.path:
    sys.path.insert(0, str(REPO_ROOT))


from functions import download_and_mat
```

There are a ton of different names for datasets you can download and use. You can find the names of all of them in the big dict inside `import_openML_data.py`.

```python
K = download_and_mat("ACSIncome", 100, theta=[1, 10000000])
print(K)
```

```python
import matplotlib.pyplot as plt
fig, ax = plt.subplots()
plt.imshow(K, cmap="viridis")
```
